```python
# Python
import numpy as np
import matplotlib.pyplot as plt
import scipy.linalg
import sklearn
```

```python
def load_iris():
    """
    Load the iris dataset and return the transposed feature matrix and labels.

    Returns:
        tuple: (X, y) where
            X (ndarray): Transposed feature matrix.
            y (ndarray): Target labels.
    """
    # Load the iris dataset and return its data and target.
    return sklearn.datasets.load_iris()['data'].T, sklearn.datasets.load_iris()['target']

# Load iris data and print it.
X, y = load_iris()
print(X, y)
```

```python
# Filter out samples with class label 0 to keep only non-zero classes.
Xs = X[:, y != 0]
ys = y[y != 0]
print(Xs, ys)
```

```python
def split_data(X, y, ratio=(2/3)):
    """
    Split the dataset into training and testing sets.

    Parameters:
        X (ndarray): Feature matrix.
        y (ndarray): Labels array.
        ratio (float, optional): Fraction of data for training (default is 2/3).

    Returns:
        tuple: (X_train, y_train, X_test, y_test)
    """
    # Set seed for reproducibility and shuffle the data
    np.random.seed(0)
    indices = np.random.permutation(X.shape[1])
    X = X[:, indices]
    y = y[indices]

    # Determine split index based on ratio
    split_idx = int(X.shape[1] * ratio)
    X_train = X[:, :split_idx]
    y_train = y[:split_idx]
    X_test = X[:, split_idx:]
    y_test = y[split_idx:]
    return X_train, y_train, X_test, y_test

# Split data into training and testing sets.
X_train, y_train, X_test, y_test = split_data(Xs, ys)
print(X_train.shape, y_train.shape, X_test.shape, y_test.shape)
```

```python
def lda_classifier(X_train, y_train, X_val, y_val):
    """
    Perform Linear Discriminant Analysis (LDA) classification.

    Computes overall and per-class means, calculates within- and between-class scatter matrices,
    solves the generalized eigenvalue problem, projects data into a lower-dimensional space,
    and determines a classification threshold.

    Parameters:
        X_train (ndarray): Training feature matrix.
        y_train (ndarray): Training labels.
        X_val (ndarray): Validation feature matrix.
        y_val (ndarray): Validation labels.

    Returns:
        tuple: (W, threshold, X_train_lda, X_val_lda, y_pred, error_rate) where
            W (ndarray): Projection vector (features x 1).
            threshold (float): Classification threshold.
            X_train_lda (ndarray): Projected training data.
            X_val_lda (ndarray): Projected validation data.
            y_pred (ndarray): Predicted labels for validation set.
            error_rate (float): Classification error rate.
    """
    # Calculate overall mean and per-class means and counts.
    overall_mean = np.mean(X_train, axis=1, keepdims=True)
    classes = np.unique(y_train)
    means = {c: np.mean(X_train[:, y_train == c], axis=1, keepdims=True) for c in classes}
    counts = {c: X_train[:, y_train == c].shape[1] for c in classes}
    N = X_train.shape[1]

    # Compute within-class scatter matrix (SW)
    SW = np.zeros((X_train.shape[0], X_train.shape[0]))
    for c in classes:
        X_c = X_train[:, y_train == c]
        X_c_centered = X_c - means[c]
        SW += X_c_centered @ X_c_centered.T
    SW = SW / N

    # Compute between-class scatter matrix (SB)
    SB = np.zeros((X_train.shape[0], X_train.shape[0]))
    for c in classes:
        diff = means[c] - overall_mean
        SB += counts[c] * (diff @ diff.T)
    SB = SB / N

    # Solve the eigenvalue problem: SB * w = lambda * SW * w
    s, U = scipy.linalg.eigh(SB, SW)
    # Order eigenvectors by descending eigenvalues and select the top one.
    W = U[:, ::-1][:, 0:1]

    # Adjust sign of W so that projected training data has a higher mean for class 2.
    X_train_lda = W.T @ X_train
    if np.mean(X_train_lda[0, y_train == 1]) > np.mean(X_train_lda[0, y_train == 2]):
        W = -W

    # Project training and validation data
    X_train_lda = W.T @ X_train
    X_val_lda = W.T @ X_val

    # Determine threshold as the average of projected class means.
    threshold = (np.mean(X_train_lda[0, y_train == 1]) + np.mean(X_train_lda[0, y_train == 2])) / 2.0

    # Classify validation samples and compute error rate.
    y_pred = np.where(X_val_lda[0] >= threshold, 2, 1)
    error_rate = np.sum(y_pred != y_val) / y_val.size

    return W, threshold, X_train_lda, X_val_lda, y_pred, error_rate

# Apply LDA classifier to training and testing data and print error rate.
W, threshold, X_train_lda, X_test_lda, y_pred, error_rate = lda_classifier(X_train, y_train, X_test, y_test)
print('Error rate:', error_rate)
```

```python
def pca_classifier(X_train, y_train, X_val, y_val, m):
    """
    Perform Principal Component Analysis (PCA) classification.

    Computes the covariance matrix, determines the top m eigenvectors,
    projects data into lower dimensions, selects a classification threshold,
    and predicts validation labels.

    Parameters:
        X_train (ndarray): Training feature matrix.
        y_train (ndarray): Training labels.
        X_val (ndarray): Validation feature matrix.
        y_val (ndarray): Validation labels.
        m (int): Number of eigenvectors to select.

    Returns:
        tuple: (P, threshold, X_train_pca, X_val_pca, y_pred, error_rate) where
            P (ndarray): Matrix of top m eigenvectors (features x m).
            threshold (float): Classification threshold.
            X_train_pca (ndarray): Projected training data.
            X_val_pca (ndarray): Projected validation data.
            y_pred (ndarray): Predicted labels for validation set.
            error_rate (float): Classification error rate.
    """
    # Compute covariance matrix of training data.
    cov = np.cov(X_train)

    # Compute eigenvalues and eigenvectors.
    s, U = np.linalg.eig(cov)

    # Select the top m eigenvectors.
    P = U[:, ::-1][:, :m]

    # Project training and validation data.
    X_train_pca = P.T @ X_train
    X_val_pca = P.T @ X_val

    # Compute threshold from training data means.
    threshold = (np.mean(X_train_pca[0, y_train == 1]) + np.mean(X_train_pca[0, y_train == 2])) / 2.0

    # Classify validation samples and calculate error rate.
    y_pred = np.where(X_val_pca[0] >= threshold, 2, 1)
    error_rate = np.sum(y_pred != y_val) / y_val.size
    return P, threshold, X_train_pca, X_val_pca, y_pred, error_rate

# Apply PCA classifier to the training and testing data.
m = 1  # Number of eigenvectors to select
P, threshold, X_train_pca, X_test_pca, y_pred, error_rate = pca_classifier(X_train, y_train, X_test, y_test, m)
print('Error rate:', error_rate)
```

```python
def plot_hist(X, y):
    """
    Plot the histogram of projected data for each class.

    Parameters:
        X (ndarray): Projected data.
        y (ndarray): Class labels.
    """
    plt.figure()
    # Plot histogram for class 1 and class 2.
    plt.hist(X[0, y == 1], bins=5, alpha=0.5, label='Versicolor', density=True)
    plt.hist(X[0, y == 2], bins=5, alpha=0.5, label='Virginica', density=True)
    plt.xlabel('Projected Data')
    plt.ylabel('Density')
    plt.title('Distribution of Projected Data')
    plt.legend()
    plt.show()

# Plot histograms for both LDA and PCA projected data.
plot_hist(X_train_lda, y_train)
plot_hist(X_test_lda, y_test)
plot_hist(X_train_pca, y_train)
plot_hist(X_test_pca, y_test)
```

```python
def estimatePCA(X, m):
    """
    Compute the top m principal components and the mean vector of X.

    Parameters:
        X (ndarray): Data matrix of shape (features x samples).
        m (int): Number of principal components to select.

    Returns:
        tuple: (P, mu) where
            P (ndarray): Matrix of top m eigenvectors (features x m).
            mu (ndarray): Mean vector of X used for centering.
    """
    mu = np.mean(X, axis=1, keepdims=True)
    Xc = X - mu
    C = np.dot(Xc, Xc.T) / X.shape[1]
    _, eigvecs = np.linalg.eigh(C)
    P = eigvecs[:, ::-1][:, :m]
    return P, mu

def applyPCA(P, X, mu):
    """
    Project the data X onto the PCA space defined by the principal components P.

    Parameters:
        P (ndarray): Principal components matrix.
        X (ndarray): Data matrix to be projected.
        mu (ndarray): Mean vector used for centering the data.

    Returns:
        ndarray: Projected data in the PCA space.
    """
    return np.dot(P.T, X - mu)

def estimateLDA(X, y):
    """
    Compute the Linear Discriminant Analysis (LDA) projection vector.

    Parameters:
        X (ndarray): Data matrix of shape (features x samples).
        y (ndarray): Class labels corresponding to the samples.

    Returns:
        ndarray: LDA projection vector (features x 1).
    """
    overall_mean = np.mean(X, axis=1, keepdims=True)
    classes = np.unique(y)
    means = {c: np.mean(X[:, y == c], axis=1, keepdims=True) for c in classes}
    counts = {c: X[:, y == c].shape[1] for c in classes}
    N = X.shape[1]

    # Compute the within-class scatter matrix (SW)
    SW = np.zeros((X.shape[0], X.shape[0]))
    for c in classes:
        Xc = X[:, y == c]
        SW += (Xc - means[c]) @ (Xc - means[c]).T
    SW /= N

    # Compute the between-class scatter matrix (SB)
    SB = np.zeros((X.shape[0], X.shape[0]))
    for c in classes:
        diff = means[c] - overall_mean
        SB += counts[c] * (diff @ diff.T)
    SB /= N

    # Solve the generalized eigenvalue problem for SB and SW
    _, U = scipy.linalg.eigh(SB, SW)
    W = U[:, ::-1][:, :1]

    # Adjust orientation so that projected data has a higher mean for class 2
    Xlda = np.dot(W.T, X)
    if np.mean(Xlda[0, y == 1]) > np.mean(Xlda[0, y == 2]):
        W = -W
    return W

def applyLDA(W, X):
    """
    Project the data X onto the LDA space using the projection vector W.

    Parameters:
        W (ndarray): LDA projection vector.
        X (ndarray): Data matrix to be projected.

    Returns:
        ndarray: Projected data in the LDA space.
    """
    return np.dot(W.T, X)

def estimateThreshold(X, y):
    """
    Estimate the classification threshold for LDA.

    Parameters:
        X (ndarray): Projected data.
        y (ndarray): Class labels.

    Returns:
        float: Classification threshold.
    """
    return (np.mean(X[0, y == 1]) + np.mean(X[0, y == 2])) / 2.0
```

```python
def pipeline():
    """
    Execute the classification pipeline using PCA and LDA.

    The pipeline performs the following steps:
      1. Uses PCA on the training data to reduce the dimensionality to 3, yielding a projection matrix.
      2. Computes the mean vector (using PCA with 1 component) for centering the data.
      3. Projects both training and test data onto the PCA space.
      4. Estimates the LDA projection vector using the PCA-transformed training data.
      5. Projects the PCA-transformed data onto the LDA space.
      6. Determines the classification threshold based on the training data.
      7. Classifies the test data using the threshold and computes the error rate.

    Returns:
        float: The classification error rate computed on the test dataset.
    """
    # Use PCA to reduce features to 3 dimensions; extract projection matrix only.
    P, _ = estimatePCA(X_train, 3)
    # Get the mean from PCA with 1 component to use for centering.
    _, mu = estimatePCA(X_train, 1)
    # Project training and testing data onto the PCA space.
    X_train_pca = applyPCA(P, X_train, mu)
    X_test_pca = applyPCA(P, X_test, mu)

    # Compute the LDA projection vector using PCA-transformed training data.
    W = estimateLDA(X_train_pca, y_train)
    # Project data onto the LDA space.
    X_train_lda = applyLDA(W, X_train_pca)
    X_test_lda = applyLDA(W, X_test_pca)

    # Estimate the classification threshold based on training data.
    threshold = estimateThreshold(X_train_lda, y_train)
    # Classify test data and compute error rate.
    y_pred = np.where(X_test_lda[0] >= threshold, 2, 1)
    error_rate = np.sum(y_pred != y_test) / y_test.size
    return error_rate

# Run the pipeline and print the error rate.
error_rate = pipeline()
print("Error rate:", error_rate)
```
