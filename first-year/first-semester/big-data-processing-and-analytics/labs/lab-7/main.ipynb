```python
import findspark
import pyspark
from jedi.api.helpers import infer
from numpy.distutils.conv_template import header
from pyspark.sql import SparkSession
from datetime import datetime


findspark.init()
sc = pyspark.SparkContext.getOrCreate()
spark = SparkSession.builder.getOrCreate()
```

```python
"""
RDD SOLUTION
"""
```

```python
registerRDD = sc.textFile("sampleData/registerSample.csv").filter(lambda x: "station" not in x)
stationsRDD = sc.textFile("sampleData/stations.csv").filter(lambda x: "id" not in x)
```

```python
criticality_threshold = 0.2
```

```python
registerRDD = registerRDD.filter(lambda x: "\t0\t0" not in x)
registerRDD.collect()
```

```python
stationsRDD.collect()
```

```python
# RDD with ((stationID, dayOfTheWeek, hour), (criticality equal to 0, 1))
mappedRDD = registerRDD.map(lambda x: ((int(x.split("\t")[0]), datetime.strptime(x.split("\t")[1], "%Y-%m-%d %H:%M:%S").strftime("%a"), datetime.strptime(x.split("\t")[1], "%Y-%m-%d %H:%M:%S").hour), (1 if int(x.split("\t")[3]) == 0 else 0, 1)))
mappedRDD.collect()
```

```python
criticalityRDD = mappedRDD.reduceByKey(lambda x, y: (x[0] + y[0], x[1] + y[1]))
criticalityRDD = criticalityRDD.map(lambda x: (x[0], x[1][0] / x[1][1]))
criticalityRDD.collect()
```

```python
criticalityRDD = criticalityRDD.filter(lambda x: x[1] > criticality_threshold)
criticalityRDD.collect()
```

```python
def select_most_critical_stations(x, y):
    if x[2] > y[2]:
        return x
    elif y[2] > x[2]:
        return y
    else:
        if x[1] > y[1]:
            return x
        elif y[1] > x[1]:
            return y
        else:
            day_list = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
            if day_list.index(x[0]) > day_list.index(y[0]):
                return x
            else:
                return y
```

```python
criticalityRDD = criticalityRDD.map(lambda x: (x[0][0], (x[0][1], x[0][2], x[1])))
mostCriticalStations = criticalityRDD.reduceByKey(select_most_critical_stations)
mostCriticalStations.collect()
```

```python
"""
SPARKSQL SOLUTION
"""
```

```python
registerDF = spark.read.load("sampleData/registerSample.csv", format="csv", header=True, inferSchema=True, sep="\t")
stationsDF = spark.read.load("sampleData/stations.csv", format="csv", header=True, inferSchema=True, sep="\t")
```

```python
registerDF.createOrReplaceTempView("register")
stationsDF.createOrReplaceTempView("stations")
```

```python
result = spark.sql("SELECT * FROM register")
result.show()
```

```python
spark.udf.register("getTimestamp",
                   lambda x: f"{x.strftime('%a')} {x.hour}",
                   "string")
spark.udf.register("criticality",
                   lambda x: 1 if x == 0 else 0,
                   "integer")
```

```python
result = spark.sql("SELECT station, getTimestamp(timestamp), SUM(criticality(free_slots))/COUNT(free_slots) AS Criticality "
                   "FROM register "
                   "WHERE EXISTS(SELECT 1 FROM register "
                   "             WHERE free_slots != 0 AND used_slots != 0)"
                   "GROUP BY station, getTimestamp(timestamp) "
                   "HAVING Criticality > 0.2")
result.show()
```
