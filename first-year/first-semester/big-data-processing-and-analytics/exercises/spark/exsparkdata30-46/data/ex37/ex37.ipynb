```python
import findspark
import pyspark
from pyspark.sql import SparkSession
```

```python
findspark.init()
sc = pyspark.SparkContext.getOrCreate()
```

```python
"""
RDD SOLUTION
"""
```

```python
inputRDD = sc.textFile("data/sensors.txt")
inputRDD.collect()
```

```python
mappedRDD = inputRDD.map(lambda x: (x.split(",")[0], float(x.split(",")[2])))
mappedRDD.collect()
```

```python
reducedRDD = mappedRDD.reduceByKey(lambda x, y: max(x, y))
reducedRDD.collect()
```

```python
"""
SPARKSQL SOLUTION
"""
```

```python
spark = SparkSession.builder.getOrCreate()
```

```python
df = spark.read.load("data/sensors.txt", format="csv", inferSchema=True, header=False, sep=",")
df.show()
```

```python
df.createOrReplaceTempView("sensors")
```

```python
spark.sql("SELECT _c0, MAX(_c2) FROM sensors "
          "GROUP BY _c0").show()
```
