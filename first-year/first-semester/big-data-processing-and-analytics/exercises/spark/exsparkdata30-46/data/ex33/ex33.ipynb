```python
import findspark
import pyspark
from pyspark.sql import SparkSession
```

```python
findspark.init()
sc = pyspark.SparkContext.getOrCreate()
```

```python
"""
RDD SOLUTION
"""
```

```python
inputRDD = sc.textFile("data/sensors.txt")
inputRDD.collect()
```

```python
mappedRDD = inputRDD.map(lambda x: float(x.split(",")[2]))
mappedRDD.collect()
```

```python
top3 = mappedRDD.top(3)
top3
```

```python
"""
SPARKSQL SOLUTION
"""
```

```python
spark = SparkSession.builder.getOrCreate()
```

```python
df = spark.read.load("data/sensors.txt", format="csv", sep=",", inferSchema=True, header=False)
df.show()
```

```python
df.createOrReplaceTempView("SENSORS")
```

```python
spark.sql("SELECT _c2 FROM SENSORS "
          "ORDER BY _C2 DESC "
          "LIMIT 3").show()
```
