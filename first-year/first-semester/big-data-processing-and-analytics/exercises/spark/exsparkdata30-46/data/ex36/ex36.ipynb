```python
import findspark
import pyspark
from pyspark.sql import SparkSession
```

```python
findspark.init()
sc = pyspark.SparkContext.getOrCreate()
```

```python
"""
RDD SOLUTION
"""
```

```python
inputRDD = sc.textFile("data/sensors.txt")
inputRDD.collect()
```

```python
avgValue = inputRDD.map(lambda x: (float(x.split(",")[2]), 1)).reduce(lambda x, y: (x[0]+y[0], x[1]+y[1]))
avgValue = avgValue[0]/avgValue[1]
avgValue
```

```python
"""
SPARKSQL SOLUTION
"""
```

```python
spark = SparkSession.builder.getOrCreate()
```

```python
df = spark.read.load("data/sensors.txt", format="csv", header=False, inferSchema=True, sep=",")
df.show()
```

```python
df.createOrReplaceTempView("sensors")
```

```python
spark.sql("SELECT AVG(_c2) FROM sensors").show()
```
