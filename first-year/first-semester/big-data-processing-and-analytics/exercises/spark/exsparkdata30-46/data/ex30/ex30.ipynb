```python
import findspark
import pyspark
from pyspark.sql import SparkSession
```

```python
findspark.init()
sc = pyspark.SparkContext.getOrCreate()
```

```python
"""
RDD SOLUTION
"""
```

```python
inputRDD = sc.textFile("data/log.txt")
inputRDD.collect()
```

```python
filteredRDD = inputRDD.filter(lambda x: "google" in x.lower())
filteredRDD.collect()
```

```python
"""
SPARKSQL SOLUTION
"""
```

```python
spark = SparkSession.builder.getOrCreate()
```

```python
df = spark.read.text("data/log.txt")
```

```python
df.createOrReplaceTempView("logs")
```

```python
filteredDF = spark.sql("SELECT * FROM logs "
                       "WHERE value LIKE '%google%'")
filteredDF.show()
```
