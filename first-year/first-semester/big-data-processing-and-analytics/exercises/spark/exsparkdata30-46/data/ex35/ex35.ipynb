```python
import findspark
import pyspark
from pyspark.sql import SparkSession
```

```python
findspark.init()
sc = pyspark.SparkContext.getOrCreate()
```

```python
"""
RDD SOLUTION
"""
```

```python
inputRDD = sc.textFile("data/sensors.txt")
inputRDD.collect()
```

```python
maxValue = inputRDD.map(lambda x: float(x.split(",")[2])).max()
maxValue
```

```python
filteredRDD = inputRDD.filter(lambda x: str(maxValue) in x).map(lambda x: x.split(",")[1])
filteredRDD.collect()
```

```python
"""
SPARKSQL SOLUTION
"""
```

```python
spark = SparkSession.builder.getOrCreate()
```

```python
df = spark.read.load("data/sensors.txt", format="csv", sep=",", inferSchema=True, header=False)
df.show()
```

```python
df.createOrReplaceTempView("SENSORS")
```

```python
spark.sql("SELECT _c1 FROM SENSORS "
          "WHERE _c2 = (SELECT MAX(_c2) FROM SENSORS)").show()
```
