```python
findspark.init()
sc = pyspark.SparkContext.getOrCreate()
spark = SparkSession.builder.getOrCreate()
```

```python
"""
RDD SOLUTION
"""
```

```python
questionsRDD = sc.textFile('data/questions.txt').map(lambda x: (x.split(",")[0], x.split(",")[2]))
questionsRDD.collect()
```

```python
answersRDD = sc.textFile('data/answers.txt').map(lambda x: (x.split(",")[1], x.split(",")[3]))#.groupByKey().map(lambda x: (x[0], list(x[1])))
answersRDD.collect()
```

```python
finalRDD = questionsRDD.cogroup(answersRDD).map(lambda x: (x[0], list(x[1][0]), list(x[1][1])))#.join(answersRDD)
finalRDD.collect()
```
