```python
import findspark
import pyspark
from pyspark.sql import SparkSession
```

```python
findspark.init()
sc = pyspark.SparkContext.getOrCreate()
```

```python
"""
RDD SOLUTION
"""
```

```python
inputRDD = sc.textFile("data/log.txt")
inputRDD.collect()
```

```python
finalRDD = inputRDD.filter(lambda x: "google" in x.lower()).map(lambda x: x.split(" - - ")[0]).distinct()
finalRDD.collect()
```

```python
"""
SPARKSQL SOLUTION
"""
```

```python
spark = SparkSession.builder.getOrCreate()
```

```python
df = spark.read.text("data/log.txt")
```

```python
df.createOrReplaceTempView("logs")
```

```python
spark.udf.register("getIP", lambda x: x.split(" - - ")[0], "string")
```

```python
selectQuery = spark.sql("SELECT getIP(*) FROM logs "
                        "WHERE value LIKE '%google%' "
                        "GROUP BY getIP(value)")
```

```python
selectQuery.show()
```
