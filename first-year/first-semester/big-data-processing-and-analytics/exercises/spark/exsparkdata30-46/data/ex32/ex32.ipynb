```python
import findspark
import pyspark
from pyspark.sql import SparkSession
```

```python
findspark.init()
sc = pyspark.SparkContext.getOrCreate()
```

```python
"""
RDD SOLUTION
"""
```

```python
inputRDD = sc.textFile("data/sensors.txt")
inputRDD.collect()
```

```python
maxPM10Value = inputRDD.map(lambda x: float(x.split(",")[2])).max()
maxPM10Value
```

```python
"""
SPARKSQL SOLUTION
"""
```

```python
spark = SparkSession.builder.getOrCreate()
```

```python
df = spark.read.load("data/sensors.txt", format="csv", sep=",", inferSchema=True, header=False)
df.show()
```

```python
df.createOrReplaceTempView("SENSORS")
```

```python
selectQuery = spark.sql("SELECT MAX(_c2) FROM SENSORS")
selectQuery.show()
```
