# Agent-Based Systems with Language Models

An agent is a system that uses a language model to reason, make decisions, and take actions using external tools. Instead of producing a single response, an agent can observe a situation, decide what to do next, and interact with resources such as databases, APIs, or search engines. This ability to reason and act in a loop allows agents to solve more complex tasks than standalone language models.
