# Step 3: Dataset Setup
## Different options
- First one is downloading using a script that places the data in the download folder (usually recommended)
- Second one is uploading the dataset to your personal/institutional Google Drive and load it from there ([Read More](https://saturncloud.io/blog/google-colab-how-to-read-data-from-my-google-drive/))
- Place the download script directly here on colab

You are free to do as you please in this phase.


```python
# !python utils/download_dataset.py
import os
import shutil

# --- Step 3.1: Download the dataset ---
print("Downloading TinyImageNet...")
!wget -q http://cs231n.stanford.edu/tiny-imagenet-200.zip
print("Download complete.")

# --- Step 3.2: Unzip into the 'data' folder ---
# The '-d data/' flag extracts into our project's 'data' folder,
# creating the 'data/tiny-imagenet-200' structure.
print("Unzipping dataset...")
!unzip -q tiny-imagenet-200.zip -d data/
print("Dataset unzipped into data/ folder.")

# --- Step 3.3: Reorganize the validation folder ---
# This script adapts the val folder to be compatible with ImageFolder.
val_dir = 'data/tiny-imagenet-200/val'
annotations_file = os.path.join(val_dir, 'val_annotations.txt')

print("Reorganizing validation directory for ImageFolder...")

try:
    # Read annotations
    with open(annotations_file, 'r') as f:
        for line in f:
            fn, cls, *_ = line.split('\t')

            # Create subfolder for the class if it doesn't exist
            class_dir = os.path.join(val_dir, cls)
            os.makedirs(class_dir, exist_ok=True)

            # Move the image into its class subfolder
            src_path = os.path.join(val_dir, 'images', fn)
            dst_path = os.path.join(class_dir, fn)

            if os.path.exists(src_path):
                shutil.move(src_path, dst_path)

    # Remove the now-empty 'images' folder and the annotations file
    shutil.rmtree(os.path.join(val_dir, 'images'))
    os.remove(annotations_file)

    print("Validation directory reorganized successfully.")

except FileNotFoundError:
    print(f"File or directory not found. Check paths: {val_dir}")
except Exception as e:
    print(f"An error occurred: {e}")
```
